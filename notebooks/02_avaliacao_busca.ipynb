# 02 — Avaliação da recuperação

Comparação entre busca semântica e híbrida, sem refazer inferências desnecessárias.

## Objetivo
Documentar os resultados de recuperação semântica e híbrida. Se existir `avaliacao/resultados_busca.json`, inspecionar sua estrutura. A tabela de métricas abaixo reproduz **valores informados no histórico da execução do projeto**; não são calculados neste notebook sem os resultados completos.

```python
from pathlib import Path
import json
RAIZ = Path.cwd()
if not (RAIZ / 'base').exists() and (RAIZ.parent / 'base').exists():
    RAIZ = RAIZ.parent
print('Raiz do projeto:', RAIZ.resolve())
```

```python
from collections import Counter
perguntas = json.loads((RAIZ / 'avaliacao' / 'perguntas.json').read_text(encoding='utf-8'))
gabarito = json.loads((RAIZ / 'avaliacao' / 'gabarito.json').read_text(encoding='utf-8'))
print('Perguntas:', len(perguntas))
print('Perguntas com gabarito de chunks:', len(gabarito))
print('Fora da base:', sum(q.get('categoria_esperada') is None for q in perguntas))
print('Categorias:', dict(Counter(q['categoria_esperada'] or 'fora_da_base' for q in perguntas)))
```

```python
# Valores registrados na execução anterior de avaliar_busca.py.
# Substituir somente por resultados reexecutados se a avaliação for refeita.
metricas_registradas = {
    'Recall@1': {'Semântica': 26.11, 'Híbrida': 31.67},
    'Recall@3': {'Semântica': 73.89, 'Híbrida': 87.78},
    'Hit@1': {'Semântica': 66.67, 'Híbrida': 80.00},
    'Hit@3': {'Semântica': 93.33, 'Híbrida': 100.00},
}
print('Métrica          Semântica   Híbrida')
for metrica, valores in metricas_registradas.items():
    print(f"{metrica:<16} {valores['Semântica']:>6.2f}%    {valores['Híbrida']:>6.2f}%")
print('\nOrigem: placar de recuperação informado na execução anterior; não recalculado aqui.')
```

```python
arq = RAIZ / 'avaliacao' / 'resultados_busca.json'
if arq.exists():
    dados = json.loads(arq.read_text(encoding='utf-8'))
    print('Arquivo de resultados encontrado.')
    print('Tipo:', type(dados).__name__)
    print('Chaves principais:', list(dados)[:12] if isinstance(dados, dict) else 'lista')
else:
    print('resultados_busca.json ainda não foi copiado para avaliacao/.')
    print('Para reproduzir as métricas, execute python app/avaliar_busca.py na raiz do projeto.')
```

## Interpretação
A busca híbrida apresentou Recall@3 de 87,78%, acima dos 73,89% da busca semântica no teste registrado. Recall@k mede a fração de chunks relevantes recuperados; Hit@k mede a proporção de perguntas com pelo menos um chunk relevante entre os k primeiros. Esses indicadores não equivalem à qualidade factual da resposta gerada.
